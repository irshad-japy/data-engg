```python
from IPython.core.display_functions import display
# import module
import pandas as pd

# Read excel file
# and store into a DataFrame
df1 = pd.read_excel('bihar_stud_result.xlsx')
df2 = pd.read_excel('up_stud_result.xlsx')

display(df1)
display(df2)
# concat both DataFrame into a single DataFrame
df = pd.concat([df1, df2])
df.reset_index(inplace=True,drop=True)
display(df)
# Export Dataframe into Excel file
df.to_excel('final_stud_result.xlsx', index=False)
```

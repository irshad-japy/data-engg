```python
import pandas as pd
# First DataFrame
df1 = pd.DataFrame({'id': ['A01', 'A02', 'A03', 'A04'],
					'Name': ['ABC', 'PQR', 'DEF', 'GHI']})
display(df1)
# Second DataFrame
df2 = pd.DataFrame({'id': ['B05', 'B06', 'B07', 'B08','B09'],
					'Name': ['XYZ', 'TUV', 'MNO', 'JKL','GHI']})
display(df2)

frames = [df1, df2]

result = pd.concat(frames)
#result = pd.concat(frames,axis=1)
display(result)
result.reset_index(inplace=True,drop=True)
display(result)
```

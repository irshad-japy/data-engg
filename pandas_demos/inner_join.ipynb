```python
import pandas as pd

df1 = pd.DataFrame({'id': ['A01', 'A02', 'A03', 'A04'],
					'Name': ['ABC', 'PQR', 'DEF', 'GHI']})
display(df1)
df2 = pd.DataFrame({'City': ['MUMBAI', 'PUNE', 'MUMBAI', 'DELHI'],
					'Age': ['12', '13', '14', '12']})
display(df2)
# the default behaviour is join='outer'
# inner join

result = pd.concat([df1, df2], axis=1, join='inner')
display(result)
```
